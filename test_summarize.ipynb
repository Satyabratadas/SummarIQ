```python
import json
from summarizer import Summarizer
```

```python
summ = Summarizer(model_path="/content/drive/MyDrive/Summar_IQ_T5Model/t5_large")
```

```python
json_path = "/content/drive/MyDrive/AI systems Project/SummarIQ/Text_extractor code/latex_extracted.json"
with open(json_path, "r") as f:
    data = json.load(f)

paper = data[0]
```

```python
# paper
```

```python
print("\n===== ABSTRACT SUMMARY =====\n")
print(summ.summarize_abstract(paper["abstract"]))

equations = paper.get("equations", [])[:5]

print("\n===== SECTION SUMMARIES =====\n")
for sec in paper["sections"]:
    print(f"\n--- {sec['name']} ---\n")
    print(summ.summarize_section(sec['name'], sec['content'], equations))

print("\n===== EQUATION EXPLANATIONS =====\n")
for i, eq in enumerate(equations, 1):
    print(f"\n--- Equation {i} ---")
    print(eq)
    print("\nExplanation:")
    print(summ.explain_equation(eq))
```
